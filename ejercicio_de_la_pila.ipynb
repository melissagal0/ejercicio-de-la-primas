```python
#Ejercicio de la prima

# 1 vamos a poner la función para comprobar si sobrevive

def sobrevive(capital, prima, siniestros):
    saldo = capital
    for siniestro in siniestros:
        saldo = saldo + prima - siniestro
        if saldo < 0:
            return False
    return True

# 2 ahora la función para hallar la prima mínima con búsqueda binaria
def prima_minima(capital, siniestros):
    if not siniestros:
        return 0

    low = 0
    high = max(siniestros)
    resultado = high

    while low <= high:
        mid = (low + high) // 2
        if sobrevive(capital, mid, siniestros):
            resultado = mid
            high = mid - 1
        else:
            low = mid + 1

    return resultado

# --- Casos de prueba ---
ejemplos = [
    (100, [80, 150, 60, 300, 90, 120]),
    (0, [50, 50, 50, 50]),
    (200, [100, 400, 50, 300]),
    (1000, [200, 250, 180, 300]),
]

# Calculamos e imprimimos el resultado
resultados = [prima_minima(capital, siniestros) for capital, siniestros in ejemplos]
print("Resultados obtenidos:", resultados)
```
